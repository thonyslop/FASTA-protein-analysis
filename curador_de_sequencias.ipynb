```python
# 1. Instalamos Biopython en silencio en esta misma celda
!pip install biopython -q

import re
from google.colab import files
from Bio import SeqIO

SEPARADOR = ""

print("👇 HAZ CLIC EN EL BOTÓN QUE APARECERÁ ABAJO PARA SUBIR TU ARCHIVO:")
uploaded = files.upload()

if not uploaded:
    print("❌ Cancelaste la subida.")
else:
    input_fasta = list(uploaded.keys())[0]
    output_fasta = "renombrado_" + input_fasta

    records_procesados = []
    modificados = 0

    for record in SeqIO.parse(input_fasta, "fasta"):
        codigo_inicial = record.id
        corchetes = re.findall(r'\[([^\]]+)\]', record.description)

        if corchetes:
            especie = corchetes[-1]
            nuevo_header = f"{codigo_inicial}{SEPARADOR}{especie}"
            record.id = nuevo_header
            record.description = ""
            record.name = nuevo_header
            modificados += 1

        records_procesados.append(record)

    SeqIO.write(records_procesados, output_fasta, "fasta")

    print(f"\n✅ ¡Listo! Se transformaron {modificados} secuencias.")
    print("⏳ Descargando el archivo resultante a tu computadora...")

    files.download(output_fasta)
```
